```python
import geopandas as gpd
import pandas as pd
import folium
from folium.features import DivIcon

# ============================================================
# 1. BACA SLS
# ============================================================

pd.set_option("display.max_columns", None)

sls = gpd.read_file(
    r"D:\peta\layout\sls\final_sls_6309_2025_1.shp"
)
sls["geometry"] = sls["geometry"].simplify(tolerance=0.0001, preserve_topology=True)

# Koordinat SHP sudah longitude / latitude
sls = sls.set_crs(
    epsg=4326,
    allow_override=True
)

# ============================================================
# 2. BUAT KECAMATAN DAN DESA
# ============================================================

kec = sls.dissolve(
    by="nmkec",
    as_index=False
)

desa = sls.dissolve(
    by=["nmkec", "nmdesa"],
    as_index=False
)

# ============================================================
# 3. CENTER
# ============================================================

bounds = sls.total_bounds

center_lon = (bounds[0] + bounds[2]) / 2
center_lat = (bounds[1] + bounds[3]) / 2

# ============================================================
# 4. MAP
# ============================================================

m = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=11,
    crs="EPSG3857",
    tiles=None,
    prefer_canvas=True
)

# ============================================================
# 5. GOOGLE STREETS
# ============================================================

folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=m&x={x}&y={y}&z={z}",
    attr="Google Streets",
    name="Google Streets",
    overlay=False,
    control=True
).add_to(m)

# ============================================================
# 6. GOOGLE SATELLITE
# ============================================================

folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=s&x={x}&y={y}&z={z}",
    attr="Google Satellite",
    name="Google Satellite",
    overlay=False,
    control=True
).add_to(m)

# ============================================================
# 7. GOOGLE LABELS
# ============================================================

folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=h&x={x}&y={y}&z={z}",
    attr="Google Labels",
    name="Google Labels",
    overlay=True,
    control=True,
    opacity=1
).add_to(m)

# ============================================================
# 8. CUSTOM PANES
#
# SLS       = PALING ATAS
# Label     = di bawah SLS
# Kecamatan = di bawah Label
# Desa      = di bawah Kecamatan
# ============================================================

folium.map.CustomPane(
    "sls",
    z_index=650
).add_to(m)

folium.map.CustomPane(
    "label",
    z_index=640
).add_to(m)

folium.map.CustomPane(
    "kecamatan",
    z_index=630
).add_to(m)

folium.map.CustomPane(
    "desa",
    z_index=620
).add_to(m)

# ============================================================
# 9. KECAMATAN — MERAH
# ============================================================

layer_kec = folium.FeatureGroup(
    name="Kecamatan",
    overlay=True,
    show=True
)

folium.GeoJson(
    kec,
    name="Batas Kecamatan",
    style_function=lambda feature: {
        "color": "red",
        "weight": 3,
        "opacity": 1,
        "fillOpacity": 0,
        "pane": "kecamatan"
    }
).add_to(layer_kec)

layer_kec.add_to(m)

# ============================================================
# 10. DESA — HIJAU
# ============================================================

layer_desa = folium.FeatureGroup(
    name="Desa",
    overlay=True,
    show=True
)

folium.GeoJson(
    desa,
    name="Batas Desa",
    style_function=lambda feature: {
        "color": "lime",
        "weight": 2,
        "opacity": 1,
        "fillOpacity": 0,
        "pane": "desa"
    }
).add_to(layer_desa)

layer_desa.add_to(m)

# ============================================================
# 11. SLS — PALING ATAS
#
# Fill cyan 20%
# Border transparan
# Klik -> kuning 15%
# ============================================================

layer_sls = folium.FeatureGroup(
    name="SLS",
    overlay=True,
    show=True
)

sls_geojson = folium.GeoJson(
    sls,
    name="SLS",
    style_function=lambda feature: {
        "color": "transparent",
        "weight": 0,
        "opacity": 0,
        "fillColor": "cyan",
        "fillOpacity": 0.20,
        "pane": "sls"
    }
)

sls_geojson.add_to(layer_sls)
layer_sls.add_to(m)

# ============================================================
# 12. CLICK SLS
#
# Cyan 20% <-> Yellow 15%
# ============================================================

sls_id = sls_geojson.get_name()

click_script = f"""
<script>
document.addEventListener("DOMContentLoaded", function() {{

    var sls = {sls_id};

    sls.eachLayer(function(layer) {{

        layer.on("click", function(e) {{

            if (layer.options.fillColor === "yellow") {{

                layer.setStyle({{
                    color: "transparent",
                    weight: 0,
                    opacity: 0,
                    fillColor: "cyan",
                    fillOpacity: 0.20
                }});

            }} else {{

                layer.setStyle({{
                    color: "transparent",
                    weight: 0,
                    opacity: 0,
                    fillColor: "yellow",
                    fillOpacity: 0.15
                }});

            }}

            L.DomEvent.stopPropagation(e);

        }});

    }});

}});
</script>
"""

m.get_root().html.add_child(
    folium.Element(click_script)
)

# ============================================================
# 13. LABEL SLS
#
# nmdesa
# nmsls [kdsubsls]
# ============================================================

label_layer = folium.FeatureGroup(
    name="Label SLS",
    overlay=True,
    show=True
)

for _, row in sls.iterrows():

    titik = row.geometry.representative_point()

    label = (
        f"{row['nmdesa']}<br>"
        f"{row['nmsls']} [{row['kdsubsls']}]"
    )

    folium.Marker(
        location=[titik.y, titik.x],
        icon=DivIcon(
            icon_size=(0, 0),
            icon_anchor=(0, 0),
            html=f"""
            <div style="
                font-size: 10px;
                font-weight: bold;
                color: white;
                white-space: nowrap;
                line-height: 1.2;
                text-shadow:
                    -1px -1px 0 black,
                     1px -1px 0 black,
                    -1px  1px 0 black,
                     1px  1px 0 black;
                pointer-events: none;
            ">
                {label}
            </div>
            """
        )
    ).add_to(label_layer)

label_layer.add_to(m)

# ============================================================
# 14. LAYER CONTROL
# ============================================================

folium.LayerControl(
    collapsed=False
).add_to(m)

# ============================================================
# 15. TAMPILKAN
# ============================================================

m
```
